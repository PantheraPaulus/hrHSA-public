```python
import pandas as pd
```

```python
pango_data = pd.read_csv("data/pangolin_data_cleaned.csv")
```

```python
top_10_pangos = pango_data.groupby("Individual_ID").size().sort_values(ascending = False)[0:9].index
```

```python
subset = pango_data[pango_data["Individual_ID"].isin(top_10_pangos)]
```

```python
subset.to_csv("data/pango_subset.csv")
```
